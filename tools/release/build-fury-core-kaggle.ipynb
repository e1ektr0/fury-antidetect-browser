# Build Fury Chromium Core for Linux x64

This Kaggle notebook builds the patched Chromium 153 core for Fury on Linux (4 vCPU, 30 GB RAM).

### Instructions:
1. Under Notebook Settings, set **Accelerator** to None (CPU), and verify **Internet** is turned **On**.
2. In Kaggle **Add-ons -> Secrets**, add a secret with label `GH_TOKEN` containing your GitHub token (or set it in cell 1).
3. Click **Save Version -> Save & Run All (Commit)**.
4. Kaggle will run the compilation in the background (~5-6 hours) and upload `fury-core-0.1.3-linux-x64.tar.xz` directly to your GitHub Release!

```python
import os

# Set GitHub Token (from Kaggle secrets or environment)
try:
    from kaggle_secrets import UserSecretsClient
    user_secrets = UserSecretsClient()
    os.environ['GH_TOKEN'] = user_secrets.get_secret('GH_TOKEN')
except Exception:
    # Alternatively, set os.environ['GH_TOKEN'] here if running locally
    pass

os.environ['TAG'] = 'v0.1.3'
print('Release target tag:', os.environ.get('TAG'))
```

```python
%%bash
set -euo pipefail

echo "=== 1. System Info ==="
lscpu | grep "Model name\|CPU(s):" || true
free -h
df -h /

echo "=== 2. Cloning Fork ==="
rm -rf fury
git clone --depth=1 https://github.com/darakcheeff/fury-antidetect-browser.git fury
cd fury

echo "=== 3. Starting Build ==="
chmod +x tools/release/build-linux-core.sh
./tools/release/build-linux-core.sh "$TAG"

echo "=== Done! Check your GitHub Releases ==="
```
